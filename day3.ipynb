## KKN - K Nearest Neighbour

```python
from sklearn.neighbors import KNeighborsClassifier
import time

knn = KNeighborsClassifier(n_neighbors=5, weights='uniform')

start = time.time()
knn.fit(X_train_scaled, y_train)
time.time()-start
```

### Neural Network (MLP - Multi Layer Perceptron)
hidden_layer_sizes -> jumlah neuron per hidden layer, alpha -> regularisasi (L2)

```python
from sklearn.neural_network import MLPClassifier

mlp = MLPClassifier(max_iter=500, early_stopping=True, random_state=42)

mlp_param = {
    'hidden_layer_sizes': [(32,), (64,), (64, 32)],
    'alpha': [1e-4, 1e-3, 1e-2],
    'learning_rate_init': [1e-3, 1e-2]
}

mlp_grid = GridSearchCV(
    estimator=mlp,
    param_grid=mlp_param,
    cv=cv,
    scoring='accuracy',
    n_jobs=-1
)

mlp_grid.fit(X_train_scaled, y_train)
```

```python
print(mlp_grid.best_params_, mlp_grid.best_score_)

best_mlp = mlp_grid.best_estimator_
accuracy_score(y_test, best_mlp.predict(X_test_scaled))
```

### Neural Network v2 - One-Hot Encoding + fitur tambahan
LabelEncoder bikin kategori jadi angka berurutan (0,1,2..) -> MLP mengira ada urutan. One-hot lebih cocok untuk neural network.
Tambah fitur marital-status & relationship (sebelumnya tidak dipakai). race & sex sengaja tidak dipakai (isu fairness).

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

num_col = ['age', 'num_edu', 'capital-gain', 'capital-loss', 'work-hour-week']
cat_col = ['workclass', 'occupation', 'country', 'marital-status', 'relationship']

X2 = df[num_col + cat_col]

# split sama persis (random_state & stratify sama) -> baris test sama dengan model lain
X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y, test_size=.2, random_state=42, stratify=y)

# scaler & encoder di dalam pipeline -> otomatis fit hanya di data train (tiap fold CV juga)
preprocess = ColumnTransformer([
    ('num', StandardScaler(), num_col),
    ('cat', OneHotEncoder(handle_unknown='ignore'), cat_col)
])
```

```python
mlp2_pipe = Pipeline([
    ('pre', preprocess),
    ('mlp', MLPClassifier(max_iter=500, early_stopping=True, random_state=42))
])

mlp2_param = {
    'mlp__hidden_layer_sizes': [(32,), (64,), (64, 32)],
    'mlp__alpha': [1e-4, 1e-3, 1e-2],
    'mlp__learning_rate_init': [1e-3, 1e-2]
}

mlp2_grid = GridSearchCV(
    estimator=mlp2_pipe,
    param_grid=mlp2_param,
    cv=cv,
    scoring='accuracy',
    n_jobs=-1
)

mlp2_grid.fit(X2_train, y2_train)
print(mlp2_grid.best_params_, mlp2_grid.best_score_)

best_mlp2 = mlp2_grid.best_estimator_
accuracy_score(y2_test, best_mlp2.predict(X2_test))
```

```python
# perbandingan adil: RFC dengan fitur & encoding yang sama
rfc2_pipe = Pipeline([
    ('pre', preprocess),
    ('rfc', RandomForestClassifier(**rfc_grid.best_params_, random_state=42, n_jobs=-1))
])
rfc2_pipe.fit(X2_train, y2_train)

for nama, m in {'MLP v2': best_mlp2, 'RFC v2': rfc2_pipe}.items():
    print('Model: ', nama)
    print(classification_report(y2_test, m.predict(X2_test)))
```

## Data Manipulation
1. add atau remove columns
2. add atau remove rows

```python
df['week'] = 'weekend'
```

```python
df.head() # menampilkan 5 data teratas
```

```python
def week (day):
    if (day == 'Sun') | (day == 'Sat'):
        return "weekend"
    else:
        return "weekday"
```

```python
week(df['day'].values[0])
```

```python
df['day'].apply(week)
```

```python
#one liner expression
df['week'] = df['day'].apply(lambda x: 'weekend' if (x == 'Sun') | (x == 'Sat') else 'weekday')
```

```python
df
```

```python
# kolom baru visitor, High -> total_bill > 17 dan size >= 3, Low
df['visitor'] = df.apply(lambda x: 'High' if x['total_bill'] > 17 and x['size'] >= 3 else 'Low', axis=1)
```

## Handling Outlier

1. IQR -> upper bound: q3 + 1.5 (iqr), lower bound: q1 - 1.5 (iqr)

```python
df['total_bill'].plot.box()
```

```python
df['total_bill'].plot()
```

```python
q1 = df['total_bill'].quantile(0.25)
q3 = df['total_bill'].quantile(0.75)
iqr = q3 - q1

print(iqr)
print('upper bounds:', q3 + 1.5 * iqr)
print('lower bounds:', q1 - 1.5 * iqr)
```

```python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/jbrownlee/Datasets/refs/heads/master/housing.csv', header=None)
```

```python
df
```

```python
X = df.iloc[:,:-1]
```

```python
y = df.iloc[:,-1]
```

```python
X
```

```python
y
```

```python
#train test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.2, random_state=42)
```

```python
X_train
```

```python
X_test
```

```python
#scaling
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Linear Regression

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr.fit(X_train_scaled, y_train)
```

```python
y_pred = lr.predict(X_test_scaled)
```

```python
y_test
```

```python
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score
```

```python
print('MAE: ', mean_absolute_error(y_test, y_pred))
print('RMSE: ', root_mean_squared_error(y_test, y_pred))
print('R2-Score: ', r2_score(y_test, y_pred))
```

## Eksperimen

model + hyperparameter tuning

```python
from sklearn.model_selection import KFold, GridSearchCV

cv = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
#SVR -> support vector regressor
from sklearn.svm import SVR

svr = SVR()

svr_param = {
    'C': [0.1,0.5,1],
    'epsilon': [0.1,0.5,1],
    'kernel': ['linear', 'rbf', 'poly']
}
```

```python
svr_grid = GridSearchCV(
    estimator=svr,
    param_grid=svr_param,
    cv=cv,
    scoring='r2',
    n_jobs=-1
)
```

```python
svr_grid.fit(X_train_scaled, y_train)
```

```python
svr_grid.best_params_
```

```python
svr_grid.best_score_
```

```python
best_svr = svr_grid.best_estimator_
```

```python
#random forrest regressor
from sklearn.ensemble import RandomForestRegressor

rfr = RandomForestRegressor()

rfr_param = {
    'n_estimators': [20,30,40,50,60],
    'max_depth': [10,15,None],
    'min_samples_split': [2,5],
    'min_samples_leaf': [1,2,5]
}
```

```python
rfr_grid = GridSearchCV(
    estimator=rfr,
    param_grid=rfr_param,
    cv=cv,
    scoring='r2',
    n_jobs=-1
)
```

```python
rfr_grid.fit(X_train_scaled, y_train)
```

```python
rfr_grid.best_score_
```

```python
rfr_grid.best_params_
```

```python
best_rfr = rfr_grid.best_estimator_
```

```python
y_pred = best_rfr.predict(X_test_scaled)
```

```python
r2_score(y_test,y_pred)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(y_test.values, label='actual')
plt.plot(y_pred, label='prediksi')

plt.legend()
```

```python
fig, ax = plt.subplots(1,2, figsize=(20,10))
ax[0].plot(y_test.values, label='actual')
ax[0].plot(best_svr.predict(X_test_scaled), label='prediksi')
ax[0].legend()
ax[0].set_title('SVR')

ax[1].plot(y_test.values, label='actual')
ax[1].plot(best_rfr.predict(X_test_scaled), label='prediksi')
ax[1].legend()
ax[1].set_title('Random Forrest')
```

```python
print('SVR')
print('MAE: ', mean_absolute_error(y_test, best_svr.predict(X_test_scaled)))
print('RMSE: ', root_mean_squared_error(y_test, best_svr.predict(X_test_scaled)))
print('R2-Score: ', r2_score(y_test, best_svr.predict(X_test_scaled)))

print('')
print('Random Forest')
print('MAE: ', mean_absolute_error(y_test, best_rfr.predict(X_test_scaled)))
print('RMSE: ', root_mean_squared_error(y_test, best_rfr.predict(X_test_scaled)))
print('R2-Score: ', r2_score(y_test, best_rfr.predict(X_test_scaled)))
```

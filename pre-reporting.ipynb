```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV, StratifiedKFold

from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

preprocessing = ColumnTransformer([
    ('num', StandardScaler(), col_num),
    ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), col_cat)
])
# sparse_output=False -> GaussianNB needs a normal (dense) array

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

models = {
    'KNN': (KNeighborsClassifier(), {
        'model__n_neighbors': [5, 15, 25],
        'model__weights': ['uniform', 'distance']
    }),
    'Decision Tree': (DecisionTreeClassifier(random_state=42, class_weight='balanced'), {
        'model__max_depth': [3, 5, 10, None],
        'model__min_samples_leaf': [1, 5, 20]
    }),
    'Random Forest': (RandomForestClassifier(random_state=42, class_weight='balanced'), {
        'model__n_estimators': [100, 200],
        'model__max_depth': [5, 10, None],
        'model__min_samples_leaf': [1, 5]
    }),
}
```

```python
best_models = {}

for name, (model, params) in models.items():
    pipe = Pipeline([('preprocessing', preprocessing), ('model', model)])

    grid = GridSearchCV(pipe, params, cv=cv, scoring='f1', n_jobs=-1)
    grid.fit(X_train, y_train)

    best_models[name] = grid.best_estimator_
    print(f'{name}: CV F1 = {grid.best_score_:.3f}  {grid.best_params_}')
```

```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

results = []
for name, model in best_models.items():
    y_pred = model.predict(X_test)
    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred),
        'Recall': recall_score(y_test, y_pred),
        'F1': f1_score(y_test, y_pred)
    })

results = pd.DataFrame(results).sort_values('F1', ascending=False)
results
```

```python
import joblib

best_name = results.iloc[0]['Model']
final_model = best_models[best_name]
print('Best model:', best_name)

from sklearn.metrics import classification_report
print(classification_report(y_test, final_model.predict(X_test)))

joblib.dump(final_model, 'churn_best_model.joblib')
```
